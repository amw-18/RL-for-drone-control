```python
from gym_pybullet_drones.envs.BaseAviary import DroneModel
from ModDSLPID import ModDSLPIDatt_torq
from TestAv2 import TestAv2PID
import matplotlib.pyplot as plt
import numpy as np
```

```python
ep_len = 2 # sec

drone_model = DroneModel.CF2X

env = TestAv2PID(drone_model=drone_model, ep_len=ep_len, freq=240)
ctrl_timestep = 1/env.SIM_FREQ

ctrl = ModDSLPIDatt_torq(drone_model=drone_model)
cumm_reward = 0
curr_rpys = []
state_history = []
computed_target_rpys = []
obs = env.reset()
for i in range(ep_len*env.SIM_FREQ):
    state = obs[:20]
    state_history.append(state)
    
    computed_target_rpy = obs[20:]
    torques = ctrl.computeControl(ctrl_timestep,
                                state,
                                computed_target_rpy
                                )
    computed_target_rpys.append(computed_target_rpy)
    curr_rpys.append(obs[7:10])
    obs, reward, done, info = env.step(torques)
    cumm_reward += reward

print(cumm_reward)
```

```python
np.sum(np.abs(np.array(computed_target_rpys) - np.array(curr_rpys)))
```
