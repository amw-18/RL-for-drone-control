```python
import gym
import os
```

```python
from stable_baselines3 import DDPG
from stable_baselines3.ddpg.policies import MlpPolicy
from stable_baselines3.common.monitor import Monitor
```

```python
mcar_env = gym.make("MountainCarContinuous-v0")

log_dir = "C:/Users/awals/OneDrive/Documents/Academic/DDP/Implementations/sb3 trials/logs"
os.makedirs(log_dir, exist_ok=True)
env = Monitor(mcar_env, log_dir)

model = DDPG(MlpPolicy, env, verbose=2, device='cpu')
```

```python
from stable_baselines3.common.callbacks import EvalCallback
eval_env = gym.make("MountainCarContinuous-v0")
callback = EvalCallback(eval_env, 
            log_path="C:/Users/awals/OneDrive/Documents/Academic/DDP/Implementations/sb3 trials/logs",
            n_eval_episodes=1,
            eval_freq=500,
            deterministic=True) 
```

```python
model.learn(6000, callback=callback)
```

```python
model.save('mcar_sb3')
```

```python
model = DDPG.load('mcar_sb3', env=env)
```

```python
obs = mcar_env.reset()
done = False
tot_rew = 0
while not done:
    action, _state = model.predict(obs)
    obs, reward, done, info = mcar_env.step(action)
    tot_rew += reward

print(tot_rew)
```
